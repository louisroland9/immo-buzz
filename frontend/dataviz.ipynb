```python
import duckdb
import pandas as pd
```

```python
# Connexion à DuckDB (en mémoire ou fichier)
con = duckdb.connect()

# Affichage propre des colonnes (pour pouvoir cliquer sur les url)
pd.set_option("display.max_colwidth", None)

# Chargement et requêtage de SQLite
con.execute("INSTALL sqlite; LOAD sqlite;")
con.execute("ATTACH '../data/app.db' AS ma_base (TYPE SQLITE);")
```

```python
# Récupération des données sous forme de DataFrame Pandas, par exemple
df = con.execute(
    "SELECT "
    "                   property_type,"
    "                   price,"
    "                   surface,"
    "                   city,"
    "                   url,"
    "                   detected_condition,"
    "                   estimated_monthly_rent,"
    "                   estimated_works_cost,"
    "                   total_cost,"
    "                   estimated_yield,"
    "                   updated_at"
    "               FROM ma_base.listing"
    "               ORDER BY estimated_yield desc"
).df()
print(df)
```

```python
# Récupération des données sous forme de DataFrame Pandas, par exemple
df = con.execute("SELECT * FROM ma_base.listing ORDER BY estimated_yield desc").df()
print(df)
```
